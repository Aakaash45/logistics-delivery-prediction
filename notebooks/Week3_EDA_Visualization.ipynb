```python
# Convert timestamp-like values into numerical hours
df["delivery_time_hours"] = (
    df["delivery_time_hours"]
    .astype(str)
    .str.extract(r'(\d+)$')[0]
    .astype(float)
)

df["expected_time_hours"] = (
    df["expected_time_hours"]
    .astype(str)
    .str.extract(r'(\d+)$')[0]
    .astype(float)
)

print(df[["delivery_time_hours", "expected_time_hours"]].head(10))
```

```python
print(df[["delivery_time_hours", "expected_time_hours"]].dtypes)
```

```python
print("Delivery time range:",
      df["delivery_time_hours"].min(),
      "to",
      df["delivery_time_hours"].max())

print("Expected time range:",
      df["expected_time_hours"].min(),
      "to",
      df["expected_time_hours"].max())
```

```python
print("Unique delivery times:",
      sorted(df["delivery_time_hours"].unique()))

print("Unique expected times:",
      sorted(df["expected_time_hours"].unique()))
```

```python
numeric_columns = [
    "distance_km",
    "package_weight_kg",
    "delivery_time_hours",
    "expected_time_hours",
    "delivery_rating",
    "delivery_cost"
]

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"{column}:")
    print(f"  Lower bound: {lower_bound:.2f}")
    print(f"  Upper bound: {upper_bound:.2f}")
    print(f"  Outliers: {len(outliers)}")
    print()
```

```python
numeric_columns = [
    "distance_km",
    "package_weight_kg",
    "delivery_time_hours",
    "expected_time_hours",
    "delivery_cost"
]

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[column] = df[column].clip(
        lower=lower_bound,
        upper=upper_bound
    )

print("Outlier treatment completed.")
```

```python
for column in numeric_columns:
    print(
        column,
        "→ Min:", df[column].min(),
        "Max:", df[column].max()
    )
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scale_columns = [
    "distance_km",
    "package_weight_kg",
    "delivery_cost"
]

df[scale_columns] = scaler.fit_transform(df[scale_columns])

print("Normalization completed.")
```

```python
print(df[scale_columns].head())
```

```python
print("Final dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```
